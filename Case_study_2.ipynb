```python
import numpy as np

def L(x):
    return 5*x - 0.2*x**2

x_values = np.linspace(0, 5, 1000)
y_values = L(x_values)

area = np.trapezoid(y_values, x_values)

print("Numerical integral:", area)
```

```python
X = df[["Study_Hours", "Attendance", "Assignment_Score"]].values
Y = df["Final_Score"].values

# Add a column of 1s for the intercept
X = np.column_stack((np.ones(len(X)), X))

print("X matrix:")
print(X)

print("\nY vector:")
print(Y)
```

```python
beta = np.linalg.inv(X.T @ X) @ X.T @ Y

print("Coefficients:")
print(beta)
```

```python
print("Intercept (b0):", beta[0])
print("Study Hours coefficient (b1):", beta[1])
print("Attendance coefficient (b2):", beta[2])
print("Assignment Score coefficient (b3):", beta[3])
```
