Task-1

```python
battery_data =  [[0, 32, 400, 20, 95],
                 [1, 35, 395, 22, 88],
                 [2, 38, 390, 25, 80],
                 [3, 40, 385, 28, 70],
                 [4, 42, 380, 30, 60],
                 [5, 45, 375, 32, 48]]
for reading in battery_data:
    print(
        "Time:", reading[0],
        "h, Temperature:", reading[1],
        "°C, Voltage:", reading[2],
        "V, Current:", reading[3],
        "A, Charge:", reading[4],
        "%"
    )
```

```python
for reading in battery_data:
    temperature = reading[1]

    if temperature > 40:
        print(reading[0], "h: High Temperature Alert")
    else:
        print(reading[0], "h: Temperature Normal")
```

```python
for reading in battery_data:
    charge = reading[4]

    if charge < 50:
        print(reading[0], "h: Low Battery Alert")
    else:
        print(reading[0], "h: Battery Charge Normal")
```

```python
def battery_status(charge):
    if charge > 70:
        return "High"
    elif 50 <= charge <= 70:
        return "Medium"
    else:
        return "Low"

for reading in battery_data:
    print(reading[0], "h:", battery_status(reading[4]))
```

```python
import sympy as sp
t = sp.symbols('t')
C = 100 - 2*t - (1/2)*t**2
dC_dt = sp.diff(C, t)
rate_2 = dC_dt.subs(t, 2)
rate_5 = dC_dt.subs(t, 5)
print("Discharge rate at t = 2:", rate_2, "% per hour")
print("Discharge rate at t = 5:", rate_5, "% per hour")
```
